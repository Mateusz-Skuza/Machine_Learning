# Porównanie metod poprawy efektywności sieci CNN na zbiorze danych CIFAR-10

Celem projektu jest porównanie wpływu różnych technik optymalizacji i regularyzacji 
na proces trenowania konwolucyjnej sieci neuronowej (CNN). Analiza jest przeprowadzona 
na zbiorze danych CIFAR-10, który zawiera 60 000 kolorowych obrazów w 10 kategoriach. 
Testowane są cztery warianty architektury: model podstawowy, model z Batch Normalization, 
model z Dropoutem oraz model łączący obie te techniki.

```python
import keras
from keras import layers, models
from keras.datasets import cifar10
import pandas as pd

# załadowanie danych
(x_train, y_train), (x_test, y_test) = cifar10.load_data()

# normalizacja danych 
x_train = x_train/255
x_test = x_test/255

# klasyczny model 
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)), 

    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    
    layers.Conv2D(64, (3, 3), activation='relu'),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax'),
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
print("\nTrenowanie: Model klasyczny")
historia_zwykly = model.fit(x_train, y_train, epochs=5, validation_data=(x_test, y_test))

# model z batch normalization
model_batch_normalization = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.BatchNormalization(),

    layers.Conv2D(64, (3, 3), activation='relu'),


    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax'),
])

model_batch_normalization.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
print("\nTrenowanie: Model z Batch Normalization")
historia_batch_normalization = model_batch_normalization.fit(x_train, y_train, epochs=5, validation_data=(x_test, y_test))

# model z dropoutem
model_dropout = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.3),

    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    
    layers.Conv2D(64, (3, 3), activation='relu'),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax'),
])

model_dropout.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
print("\nTrenowanie: Model z Dropoutem")
historia_dropout = model_dropout.fit(x_train, y_train, epochs=5, validation_data=(x_test, y_test))

# model z batch normalization i dropoutem 
model_batch_normalization_dropout = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.3),

    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.BatchNormalization(),

    layers.Conv2D(64, (3, 3), activation='relu'),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax'),
])

model_batch_normalization_dropout.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
print("\nTrenowanie: Model z Batch Normalization i Dropoutem")
historia_batch_normalization_dropout = model_batch_normalization_dropout.fit(x_train, y_train, epochs=5, validation_data=(x_test, y_test))

wyniki = {
    "Model": ["Klasyczny", "Batch Normalization", "Dropout", "Batch Normalization + Dropout"],
    "Strata": [
        historia_zwykly.history['val_loss'][-1],
        historia_batch_normalization.history['val_loss'][-1],
        historia_dropout.history['val_loss'][-1], 
        historia_batch_normalization_dropout.history['val_loss'][-1]],

    "Dokładność":  [
        f"{historia_zwykly.history['val_accuracy'][-1]* 100:.2f}%",
        f"{historia_batch_normalization.history['val_accuracy'][-1]* 100:.2f}%", 
        f"{historia_dropout.history['val_accuracy'][-1]* 100:.2f}%", 
        f"{historia_batch_normalization_dropout.history['val_accuracy'][-1]* 100:.2f}%"]
}

tabelka = pd.DataFrame(wyniki)
print("\n Porównanie wyników (ostatnia epoka)")
print(tabelka.to_string(index=False))
```

#### WNIOSKI NA PODSTAWIE EKSPERYMENTÓW:
- Batch Normalization okazał się najskuteczniejszy, osiągając najwyższą dokładność (67.43%) i najniższą stratę (0.9601) w krótkim czasie 5 epok.
- Dropout w tak krótkim procesie trenowania obniżył skuteczność modelu (63.32%), co wynika z faktu, że regularyzacja ta wymaga zazwyczaj większej liczby epok do pełnej zbieżności.
- Połączenie BN i Dropoutu przy 5 epokach doprowadziło do niedouczenia (56.92%), co sugeruje, że zbyt silna regularyzacja spowalnia proces uczenia przy małej liczbie iteracji.
- Dodanie Batch Normalization znacząco zwiększyło koszt obliczeniowy,wydłużając czas trwania epoki o około 60% względem modelu klasycznego.
